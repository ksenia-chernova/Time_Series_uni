# Задание 2
- Построить графики АЧХ и ФЧХ для сигналов (тремор) по вариантам.

Сигнал - data\lab5_data6.txt

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

signal_df = pd.read_csv("data/lab5_data6.txt", sep=';', usecols=['отсчеты', 'Канал0', 'Канал1'])

samples = signal_df['отсчеты'].values
ch0 = signal_df['Канал0'].values
ch1 = signal_df['Канал1'].values

N = len(ch0)
fs = 100.0
t = samples / fs

fig, axes = plt.subplots(2, 1, figsize=(12, 6))

axes[0].plot(t, ch0)
axes[0].set_title('Канал 0: исходный сигнал')
axes[0].set_xlabel('Время (с)')
axes[0].set_ylabel('Амплитуда')
axes[0].grid(True)

axes[1].plot(t, ch1, 'r-')
axes[1].set_title('Канал 1: исходный сигнал')
axes[1].set_xlabel('Время (с)')
axes[1].set_ylabel('Амплитуда')
axes[1].grid(True)

plt.tight_layout()
```

```python
X0_fft = np.fft.fft(ch0)
freqs = np.fft.fftfreq(N, d=1/fs)

# АЧХ и ФЧХ
magnitude = np.abs(X0_fft)
phase = np.angle(X0_fft)  # фаза в радианах

fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# АЧХ
axes[0].stem(freqs, magnitude)
axes[0].set_title('АЧХ: Амплитудно-частотная характеристика (Канал 0)')
axes[0].set_xlabel('Частота (Гц)')
axes[0].set_ylabel('Модуль $|X(f)|$')
axes[0].grid(True)

# ФЧХ
axes[1].stem(freqs, phase)
axes[1].set_title('ФЧХ: Фазо-частотная характеристика (Канал 0)')
axes[1].set_xlabel('Частота (Гц)')
axes[1].set_ylabel('Фаза (рад)')
axes[1].grid(True)

plt.tight_layout()
```

```python
X1_fft = np.fft.fft(ch1)
freqs = np.fft.fftfreq(N, d=1/fs)

# АЧХ и ФЧХ
magnitude = np.abs(X1_fft)
phase = np.angle(X1_fft)  # фаза в радианах

fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# АЧХ
axes[0].stem(freqs, magnitude)
axes[0].set_title('АЧХ: Амплитудно-частотная характеристика (Канал 1)')
axes[0].set_xlabel('Частота (Гц)')
axes[0].set_ylabel('Модуль $|X(f)|$')
axes[0].grid(True)

# ФЧХ
axes[1].stem(freqs, phase)
axes[1].set_title('ФЧХ: Фазо-частотная характеристика (Канал 1)')
axes[1].set_xlabel('Частота (Гц)')
axes[1].set_ylabel('Фаза (рад)')
axes[1].grid(True)

plt.tight_layout()
```
